```python
from google.colab import userdata
import os
os.environ["OPENAI_API_KEY"] = userdata.get('OPENAI_API_KEY')
# AI_MODEL = userdata.get('OPENAI_MODEL_NAME')
AI_MODEL = 'gpt-5'
```

```python
system_prompt="""
You are an expert at structured data extraction. You will be given unstructured
text from a news article and should convert it into the given structure.
"""


news_article="""
With an estimated $950 million jackpot, Monday’s Powerball draw will be the 10th
largest in the game’s history after no player matched all six number in
Saturday’s draw, the Multi-State Lottery Association said in a press release.
"""
```

```python
from pydantic import BaseModel
from openai import OpenAI

client = OpenAI()

class NewsExtraction(BaseModel):
    lottery: str
    jackpot: int
    organizer: str

completion = client.chat.completions.parse(
    model=AI_MODEL,
    messages=[
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": news_article}
    ],
    response_format=NewsExtraction,
)

news_information = completion.choices[0].message.parsed
```

```python
lottery = news_information.lottery
jackpot = news_information.jackpot
organizer = news_information.organizer
print(f"{lottery=}")
print(f"{jackpot=}")
print(f"{organizer=}")
```

# With langchain

```python
!pip install openai langchain langchain_openai python-dotenv toml pydantic --quiet
```

```python
from google.colab import userdata
import os
os.environ["OPENAI_API_KEY"] = userdata.get('OPENAI_API_KEY')
AI_MODEL = userdata.get('OPENAI_MODEL_NAME')
```

```python
system_prompt="""
You are an expert at structured data extraction. You will be given unstructured
text from a news article and should convert it into the given structure.
"""


news_article="""
With an estimated $750 million jackpot, Monday’s Powerball draw will be the 10th
largest in the game’s history after no player matched all six number in
Saturday’s draw, the Multi-State Lottery Association said in a press release.
"""
```

```python
from pydantic import BaseModel
from langchain_openai import ChatOpenAI
from langchain_core.messages import AIMessage, HumanMessage, SystemMessage

client = ChatOpenAI(model=AI_MODEL)

class NewsExtraction(BaseModel):
    lottery: str
    jackpot: int
    organizer: str

messages=[
    SystemMessage(system_prompt),
    HumanMessage(news_article)
]

news_information = client.with_structured_output(NewsExtraction).invoke(messages)
```

```python
lottery = news_information.lottery
jackpot = news_information.jackpot
organizer = news_information.organizer
print(f"{lottery=}")
print(f"{jackpot=}")
print(f"{organizer=}")
```
